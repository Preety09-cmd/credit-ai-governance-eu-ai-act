```python
## Explainability using LIME

LIME explains individual predictions by approximating the model locally.

This helps understand why a specific customer was classified as high or low risk.
```

```python
from lime.lime_tabular import LimeTabularExplainer
import numpy as np

explainer = LimeTabularExplainer(
    training_data=np.array(X_train),
    feature_names=X_train.columns.tolist(),
    class_names=["Low Risk", "High Risk"],
    mode="classification"
)

exp = explainer.explain_instance(
    np.array(X_test.iloc[0]),
    rf_model.predict_proba
)

exp.show_in_notebook()
```

## Governance Insights

This prototype demonstrates how AI systems in financial services can be made more trustworthy:

- Explainability (SHAP, LIME) improves transparency of model decisions
- Model comparison supports governance review
- Audit logs enable traceability of decisions
- Monitoring helps detect abnormal model behavior

This aligns with the EU AI Act requirements for high-risk AI systems, such as credit scoring.
